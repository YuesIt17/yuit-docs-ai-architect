```python
import subprocess
import sys
from pathlib import Path

REPO_URL = "https://github.com/YuesIt17/yuit-docs-ai-architect.git"
REPO_BRANCH = "hw-3-multi-agent-rag"  # after merging into main, change to "main"

if "google.colab" in sys.modules:
    repo_root = Path("/content/yuit-docs-ai-architect")
    if not (repo_root / "hw-3/code/shopping_assistant").exists():
        subprocess.run(
            ["git", "clone", "-q", "--branch", REPO_BRANCH, "--depth", "1", REPO_URL, str(repo_root)],
            check=True,
        )
    code_root = repo_root / "hw-3/code"
else:
    code_root = Path("..").resolve()
    repo_root = code_root.parents[1]

requirements = repo_root / "tools/python/requirements.txt"
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(requirements)], check=True)
sys.path.insert(0, str(code_root))

from shopping_assistant import MESSAGES, USER_QUERY, build_app, clear_messages
```

```python
clear_messages()
app = build_app()
result = app.invoke(
    {
        "user_query": USER_QUERY,
        "constraints": {},
        "policy_chunks": [],
        "policy_verdict": "",
        "cited_policy_ids": [],
        "candidates": [],
        "cart": [],
        "promo_notes": [],
        "final_answer": "",
    }
)

print("\n=== FINAL ANSWER ===")
print(result["final_answer"])
print("\n=== MESSAGE LOG ===")
for msg in MESSAGES:
    print(msg)
```
